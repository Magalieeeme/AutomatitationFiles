```python
import tkinter as tk
from tkinter import filedialog
import pandas as pd

# Función para seleccionar archivos de Excel y procesarlos
def seleccionar_archivos():
    archivos = filedialog.askopenfilenames(title="Seleccionar archivos Excel", filetypes=[("Archivos Excel", "*.xlsx")])
    
    if not archivos:
        return
    
    datos = []
    
    for archivo in archivos:
        try:
            # Leer el archivo Excel
            df = pd.read_excel(archivo, header=None)
            df.columns = df.iloc[0]  # Establece la primera fila como encabezados
            
            # Obtener las celdas C4 y F1
            c4 = df.iloc[3, 2] if df.shape[0] > 3 and df.shape[1] > 2 else None
            d1 = df.iloc[0, 3]
            f1 = df.iloc[0, 5] if df.shape[1] > 5 else None
            
            datos.append({'Archivo': archivo, 'C4': c4,'D1': d1, 'F1': f1})
        except Exception as e:
            print(f"Error al procesar el archivo {archivo}: {str(e)}")
    
    if datos:
        # Crear un DataFrame con los resultados
        resultado_df = pd.DataFrame(datos)

        # Verificar si todas las cadenas de la columna "D1" son iguales
        if resultado_df['D1'].nunique() == 1:
            print("Todos tienen la misma fecha")
        else:
            print("Revisar archivo en las fechas que lo descargaste")

        # Verificar si todas las cadenas de F1 son iguales
        if resultado_df['F1'].nunique() == 1:
            print("Todas tiene la duración correcta")
        else:
            print("Revisar el archivo en tiempo que lo descargaste")
        # Guardar el DataFrame en un archivo Excel llamado "procesamiento.xlsx"
        resultado_df.to_excel("procesamiento.xlsx", index=False)
        
        print("Datos procesados y guardados en 'procesamiento.xlsx'")
        
        # Reemplazar "M1" a "M9" por "M01" a "M09" en la columna B
        reemplazar_m1_to_m9("procesamiento.xlsx")
    else:
        print("No se encontraron datos para procesar.")
    
# Función para reemplazar "M1" a "M9" por "M01" a "M09" en la columna B
def reemplazar_m1_to_m9(archivo):
    try:
        df = pd.read_excel(archivo)
        df['C4'] = df['C4'].str.replace('M0', 'M000')
        df['C4'] = df['C4'].str.replace('M1', 'M01')
        df['C4'] = df['C4'].str.replace('M2', 'M02')
        df['C4'] = df['C4'].str.replace('M3', 'M03')
        df['C4'] = df['C4'].str.replace('M4', 'M04')
        df['C4'] = df['C4'].str.replace('M5', 'M05')
        df['C4'] = df['C4'].str.replace('M6', 'M06')
        df['C4'] = df['C4'].str.replace('M7', 'M07')
        df['C4'] = df['C4'].str.replace('M8', 'M08')
        df['C4'] = df['C4'].str.replace('M9', 'M09')
        df['C4'] = df['C4'].str.replace('M010', 'M10')
        df['C4'] = df['C4'].str.replace('M011', 'M11')
        df['C4'] = df['C4'].str.replace('M012', 'M12')
        df['C4'] = df['C4'].str.replace('M013', 'M13')
        df['C4'] = df['C4'].str.replace('M014', 'M14')
        df['C4'] = df['C4'].str.replace('M015', 'M15')
        df['C4'] = df['C4'].str.replace('M016', 'M16')
        df['C4'] = df['C4'].str.replace('M017', 'M17')
        df['C4'] = df['C4'].str.replace('M018', 'M18')
        df['C4'] = df['C4'].str.replace('M019', 'M19')
        df['C4'] = df['C4'].str.replace('M020', 'M20')
        df['C4'] = df['C4'].str.replace('M021', 'M21')
        df['C4'] = df['C4'].str.replace('M022', 'M22')
        df['C4'] = df['C4'].str.replace('M023', 'M23')

        df.to_excel(archivo, index=False)
        # Extraer caracteres de C4 desde la cuarta letra hasta la sexta letra y colocarlos en la columna E
        df['Campus'] = df['C4'].str[3:5]

        # Contar cuántas veces aparecen "C1" y "C2" en la columna D
        contador_c1 = (df['Campus'] == 'C1').sum()
        contador_c2 = (df['Campus'] == 'C2').sum()
        contador_rec = (df['Campus'] == '-R').sum()
        contador_prope = (df['Campus'] == '0C').sum()

        print(f"Cantidad de 'C1' en columna D: {contador_c1}")
        print(f"Cantidad de 'C2' en columna D: {contador_c2}")
        print(f"Cantidad de 'Recursamiento' en columna D: {contador_rec}")
        print(f"Cantidad de 'Propedeuticos' en columna D: {contador_prope}")

        df.to_excel(archivo, index=False)
        print(f"Reemplazo, extracción y conteo en {archivo} completados.")
    except Exception as e:
        print(f"Error al reemplazar, extraer y contar en {archivo}: {str(e)}")

# Llamar a la función para seleccionar archivos
seleccionar_archivos()
```
