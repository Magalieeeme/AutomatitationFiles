```python
import os

def crear_carpetas(rango_inicial, rango_final, ruta_base):
    for numero in range(rango_inicial, rango_final + 1):
        nombre_carpeta = str(numero)
        ruta_carpeta = os.path.join(ruta_base, nombre_carpeta)

        # Verificar si la carpeta ya existe antes de crearla
        if not os.path.exists(ruta_carpeta):
            os.makedirs(ruta_carpeta)
            print(f'Carpeta creada: {ruta_carpeta}')
        else:
            print(f'La carpeta {ruta_carpeta} ya existe')

# Especifica el rango de números y la ruta base
rango_inicial = 24
rango_final = 26
ruta_base = 'C:\\Users\\magal\\Desktop'  

crear_carpetas(rango_inicial, rango_final, ruta_base)
```
